```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle

def plot_phasor(magnitude, angle_deg, color='red', label=''):
    """Plot a single phasor"""
    angle_rad = np.deg2rad(angle_deg)
    x = magnitude * np.cos(angle_rad)
    y = magnitude * np.sin(angle_rad)
    
    plt.quiver(0, 0, x, y, angles='xy', scale_units='xy', scale=1,
               color=color, width=0.01, label=label)
    return x, y

# Create figure
plt.figure(figsize=(8, 8))

# Plot multiple phasors
phasors = [
    (4, 90, 'red', 'U2'),
    (7, 0, 'blue', 'U1'),]

for mag, angle, color, label in phasors:
    plot_phasor(mag, angle, color, label)

# Add unit circle
circle = Circle((0, 0), 1, fill=False, linestyle='--', alpha=0.3)
plt.gca().add_patch(circle)

# Formatting

max_mag = max([p[0] for p in phasors])
plt.xlim(-2,8)
plt.ylim(-2, 5)
plt.axhline(y=0, color='k', alpha=0.3)
plt.axvline(x=0, color='k', alpha=0.3)
plt.legend()
plt.title('Phasor Diagram')
plt.gca().set_aspect('equal')

plt.show()
```
