```python
#IMPORTS
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize, fsolve
import holoviews as hv
hv.extension('bokeh')
```

```python
#ALGEMEENHEDEN
data = np.loadtxt(fname = "Helium.txt", delimiter = "\t", dtype = str).T
data = np.char.replace(data, ',', '.').astype(float)
pos = data[0] #[mm]
I = data[1] #[arb. eenh.]
sigma1_I = np.sqrt(I)
```

```python
#SCATTERPLOT I (met fout)
def scatterplot_I(positie, intensiteit, fout_int):
    ax.scatter(positie, intensiteit, color = "black", label = "Experimenteel gemeten intensiteit")
    ax.errorbar(positie, intensiteit, yerr = fout_int, color = 'black', marker = "o", fmt = " ", capsize=2, capthick=0.6, linewidth=0.6)
    ax.set_xlabel("positie x [m]")
    ax.set_ylabel("Intensiteit I [arb. eenh.]")

fig, ax = plt.subplots(figsize = (10, 5))
scatterplot_I(pos, I, sigma1_I)
ax.legend()
```

```python
#BEGINPARAMETERS AFSCHATTEN ADHV INTERACTIEVE PLOT

def intensiteit(x, x0, gamma, A, y0):
    I = (A / np.pi) * (gamma) / ((x - x0)**2 + gamma**2) + y0
    return I

def model_curve(x0, gamma, A, y0):
    """Speciale functie die geupdated kan worden met sliders."""
    x = np.linspace(np.min(pos)*1e3, np.max(pos)*1e3, 1000)
    return hv.Curve([(x_i, intensiteit(x_i, x0, gamma, A, y0)) for x_i in x]).opts(color='red', line_dash='dashed')

dmap = hv.DynamicMap(model_curve, kdims=['x0', 'gamma', 'A', 'y0']) \
    .redim.values(x0 = np.arange(0, 1e-3, 0.01e-3), gamma = np.arange(0, 5e-3, 0.1e-3), A = np.arange(0, 1, 0.01), y0 = np.arange(0, 101, 1))
dmap * \
    hv.Scatter([(x, y) for x, y in zip(pos, I)]) \
        .options(xlabel="x [m]", ylabel="I [arb. eenh.]", xlim=(-42e-3, 42e-3), size=6, width=500, height=350, color="black") * \
    hv.ErrorBars([(x, y, dy) for x, y, dy in zip(pos, I, sigma1_I)])
```

```python
#chisq
from scipy.stats import chi2

nu = len(I) - 4 #4 parameters afschatten

def chi_sq(pos, I, x0, gamma, A, y0, fout_I):
    chi_squared = np.sum((I - intensiteit(pos, x0, gamma, A, y0))**2 / (fout_I)**2)
    return chi_squared

#beginwaardes fit - worden constant gefinetuned tot tevreden
x0 = 728
gamma = 2.5
A = 2200
y0 = 0
```

```python
#x0

def chisq_x0(pos, I, waarde, gamma, A, y0, fout_I):
    waardes = np.linspace(waarde - 1, waarde + 1, 1000)
    chi_waardes = []
    for i in range(len(waardes)):
        chi_waarde = chi_sq(pos, I, waardes[i], gamma, A, y0, fout_I)
        chi_waardes.append(chi_waarde)
    return (waardes, chi_waardes)

x0_waardes = chisq_x0(pos, I, x0, gamma, A, y0, sigma1_I)[0]
chisq_x0_waardes = chisq_x0(pos, I, x0, gamma, A, y0, sigma1_I)[1]

fig, ax = plt.subplots(figsize = (8, 5))
ax.plot(x0_waardes, chisq_x0_waardes, color = "blue")
ax.set_xlabel(r"verschuivingsparameter $x_0$ [m]")
ax.set_ylabel(r"$\chi^2$")

chisq_min_x0 = np.min(chisq_x0_waardes)
min_x0 = x0_waardes[np.argmin(chisq_x0_waardes)]
print("Minimale x0 =", min_x0)

conf_lvl_x0 = chisq_min_x0 + chi2.ppf(0.68, df = 4)
ax.axhline(y = conf_lvl_x0, color = "blue")

#LINEAIRE INTERPOLATIE VOOR x0-WAARDE VAN SNIJPUNTEN
snijpunten_x0 = []
for i in range(len(chisq_x0_waardes)-1):
    if (chisq_x0_waardes[i] - conf_lvl_x0) * (chisq_x0_waardes[i+1] - conf_lvl_x0) < 0:
        x1, x2 = x0_waardes[i], x0_waardes[i+1]
        y1, y2 = chisq_x0_waardes[i] - conf_lvl_x0, chisq_x0_waardes[i+1] - conf_lvl_x0
        x_intersect = x1 - y1 * (x2 - x1) / (y2 - y1)
        snijpunten_x0.append(x_intersect)

print("sigma1 links =", np.abs(snijpunten_x0[0] - min_x0), "sigma1 rechts =", np.abs(min_x0 - snijpunten_x0[1]))
print(snijpunten_x0)
plt.savefig("x0_waardes.png")
```

```python
#gamma

def chisq_gamma(pos, I, x0, waarde, A, y0, fout_I):
    waardes = np.linspace(waarde - 2, waarde + 1, 1000)
    chi_waardes = []
    for i in range(len(waardes)):
        chi_waarde = chi_sq(pos, I, x0, waardes[i], A, y0, fout_I)
        chi_waardes.append(chi_waarde)
    return (waardes, chi_waardes)

gamma_waardes = chisq_gamma(pos, I, min_x0, gamma, A, y0, sigma1_I)[0]
chisq_gamma_waardes = chisq_gamma(pos, I, min_x0, gamma, A, y0, sigma1_I)[1]

fig, ax = plt.subplots(figsize = (8, 5))
ax.plot(gamma_waardes, chisq_gamma_waardes, color = "blue")
ax.set_xlabel(r"schaalparameter $\gamma$")
ax.set_ylabel(r"$\chi^2$")

chisq_min_gamma = np.min(chisq_gamma_waardes)
min_gamma = gamma_waardes[np.argmin(chisq_gamma_waardes)]
print("Minimale gamma =", min_gamma)

conf_lvl_gamma = chisq_min_gamma + chi2.ppf(0.68, df = 4)
ax.axhline(y = conf_lvl_gamma, color = "blue")

#LINEAIRE INTERPOLATIE VOOR gamma-WAARDE VAN SNIJPUNTEN
snijpunten_gamma = []
for i in range(len(chisq_gamma_waardes)-1):
    if (chisq_gamma_waardes[i] - conf_lvl_gamma) * (chisq_gamma_waardes[i+1] - conf_lvl_gamma) < 0:
        x1, x2 = gamma_waardes[i], gamma_waardes[i+1]
        y1, y2 = chisq_gamma_waardes[i] - conf_lvl_gamma, chisq_gamma_waardes[i+1] - conf_lvl_gamma
        x_intersect = x1 - y1 * (x2 - x1) / (y2 - y1)
        snijpunten_gamma.append(x_intersect)

print("sigma1 links =", np.abs(snijpunten_gamma[0] - min_gamma), "sigma1 rechts =", np.abs(min_gamma - snijpunten_gamma[1]))
print(snijpunten_gamma)
plt.savefig("gamma_waardes.png")
```

```python
#A

def chisq_A(pos, I, x0, gamma, waarde, y0, fout_I):
    waardes = np.linspace(waarde - 5000, waarde + 100, 1000)
    chi_waardes = []
    for i in range(len(waardes)):
        chi_waarde = chi_sq(pos, I, x0, gamma, waardes[i], y0, fout_I)
        chi_waardes.append(chi_waarde)
    return (waardes, chi_waardes)

A_waardes = chisq_A(pos, I, min_x0, min_gamma, A, y0, sigma1_I)[0]
chisq_A_waardes = chisq_A(pos, I, min_x0, min_gamma, A, y0, sigma1_I)[1]

fig, ax = plt.subplots(figsize = (8, 5))
ax.plot(A_waardes, chisq_A_waardes, color = "blue")
ax.set_xlabel(r"vermenigvuldigingsfactor $A$")
ax.set_ylabel(r"$\chi^2$")

chisq_min_A = np.min(chisq_A_waardes)
min_A = A_waardes[np.argmin(chisq_A_waardes)]
print("Minimale A =", min_A)

conf_lvl_A = chisq_min_A + chi2.ppf(0.68, df = 4)
ax.axhline(y = conf_lvl_A, color = "blue")

#LINEAIRE INTERPOLATIE VOOR A-WAARDE VAN SNIJPUNTEN
snijpunten_A = []
for i in range(len(chisq_A_waardes)-1):
    if (chisq_A_waardes[i] - conf_lvl_A) * (chisq_A_waardes[i+1] - conf_lvl_A) < 0:
        x1, x2 = A_waardes[i], A_waardes[i+1]
        y1, y2 = chisq_A_waardes[i] - conf_lvl_A, chisq_A_waardes[i+1] - conf_lvl_A
        x_intersect = x1 - y1 * (x2 - x1) / (y2 - y1)
        snijpunten_A.append(x_intersect)

print("sigma1 links =", np.abs(snijpunten_A[0] - min_A), "sigma1 rechts =", np.abs(min_A - snijpunten_A[1]))
print(snijpunten_A)
plt.savefig("A_waardes.png")
```

```python
#y0

def chisq_y0(pos, I, x0, gamma, A, waarde, fout_I):
    waardes = np.linspace(waarde - 5, waarde + 5, 1000)
    chi_waardes = []
    for i in range(len(waardes)):
        chi_waarde = chi_sq(pos, I, x0, gamma, A, waardes[i], fout_I)
        chi_waardes.append(chi_waarde)
    return (waardes, chi_waardes)
y0_waardes = chisq_y0(pos, I, min_x0, min_gamma, min_A, y0, sigma1_I)[0]
chisq_y0_waardes = chisq_y0(pos, I, min_x0, min_gamma, min_A, y0, sigma1_I)[1]

fig, ax = plt.subplots(figsize = (8, 5))
ax.plot(y0_waardes, chisq_y0_waardes, color = "blue")
ax.set_xlabel(r"offset $y_0$")
ax.set_ylabel(r"$\chi^2$")

chisq_min_y0 = np.min(chisq_y0_waardes)
min_y0 = y0_waardes[np.argmin(chisq_y0_waardes)]
print("Minimale y0 =", min_y0)

conf_lvl_y0 = chisq_min_y0 + chi2.ppf(0.68, df = 4)
ax.axhline(y = conf_lvl_y0, color = "blue")

#LINEAIRE INTERPOLATIE VOOR y0-WAARDE VAN SNIJPUNTEN
snijpunten_y0 = []
for i in range(len(chisq_y0_waardes)-1):
    if (chisq_y0_waardes[i] - conf_lvl_y0) * (chisq_y0_waardes[i+1] - conf_lvl_y0) < 0:
        x1, x2 = y0_waardes[i], y0_waardes[i+1]
        y1, y2 = chisq_y0_waardes[i] - conf_lvl_y0, chisq_y0_waardes[i+1] - conf_lvl_y0
        x_intersect = x1 - y1 * (x2 - x1) / (y2 - y1)
        snijpunten_y0.append(x_intersect)

print("sigma1 links =", np.abs(snijpunten_y0[0] - min_y0), "sigma1 rechts =", np.abs(min_y0 - snijpunten_y0[1]))
print(snijpunten_y0)
plt.savefig("y0_waardes.png")
```

```python
#BESTE FIT VOLGENS ML-FIT
pos_fit = np.linspace(np.min(pos), np.max(pos), 1000)
#scatterplot
fig, ax = plt.subplots(figsize = (10, 5))
scatterplot_I(pos, I, sigma1_I)
#fitplot
ax.plot(pos_fit, intensiteit(pos_fit, min_x0, min_gamma, min_A, min_y0), 'r--', label = "model")
ax.legend()
plt.savefig("Lorentzfit.png")
```
