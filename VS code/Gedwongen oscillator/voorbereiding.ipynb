```python
#faseverschil
freq = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
b = np.linspace(0.1, 1, 10)

def plot_b_freq(freq, b, col, index):
    fase = - np.arctan2(np.sqrt(2) * b * freq, 1 - freq**2)
    ax.plot(freq, fase, color = col, label = f"b = {index}")

fig, ax = plt.subplots()
cols = ["red", "orange", "yellow", "green", "blue", "purple", "pink", "brown", "grey", "black"]
for i in range(len(b)):
    plot_b_freq(freq, b[i], cols[i], np.round(b[i], 1))

ax.set_title(r"Fasespectrum i.f.v. $b$")
ax.set_xlabel(r"Dimensieloze frequentie $w = \frac{\omega}{\omega_0}$ [rad / s]")
ax.set_ylabel(r"Faseverschil $\varphi(w)$ [rad]")
ax.legend()
```
