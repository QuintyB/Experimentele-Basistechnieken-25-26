```python

a_deg = 60 + 18/60       
d_deg = 35 + 46/60       
sigma_a_deg = 10/60      
sigma_d_deg = 20/60     


a = np.deg2rad(a_deg)
d = np.deg2rad(d_deg)
sigma_a = np.deg2rad(sigma_a_deg)
sigma_d = np.deg2rad(sigma_d_deg)


n = np.sin((a + d)/2) / np.sin(a/2)


dn_da = 0.5*(np.sin(a/2)*np.cos((d+a)/2)-np.sin((d+a)/2)*np.cos(a/2))/((np.sin(a/2))**2)
dn_dd = 0.5 * np.cos((a + d)/2) / np.sin(a/2)

sigma_n = np.sqrt((dn_da * sigma_a)**2 + (dn_dd * sigma_d)**2)
print(n)
print(sigma_n)
```
