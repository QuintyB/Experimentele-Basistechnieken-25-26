```python
import numpy as np
import sympy as sp
```

```python
#opdracht 2 - redoxtitratie UwU
#getritreerd volume
def fout_conc(n, V, dn, dV):
    N, v = sp.symbols('N v')
    c = N / v
    s_func = sp.sqrt((sp.diff(c, N))**2 * dn**2 + (sp.diff(c, v))**2 * dV**2)
    s = s_func.subs([(N, n), (v, V)])
    return s

Vb = np.array([33.2, 23.6, 29.35])
Ve = np.array([43.55, 33.90, 38.90])
dV = np.sqrt(0.05**2 + 0.05**2)
Vtit_list = np.abs(Vb - Ve)

Vtit = np.mean(Vtit_list)
stdev = np.sqrt(1 / 2 * np.std(Vtit_list)**2)
print(Vtit, stdev/sp.N(sp.sqrt(3)))

nI = 0.005 * Vtit*1e-3 #geen fout op conc. I2 #mL
dnI = 0.005 * stdev*1e-3 #mL

print(nI, dnI)

nS = nI #1:1 verhouding
Vwijn = 20e-3
dwijn = 0.02e-3
cS = nS / Vwijn
dnS = dnI
dcS = fout_conc(nS, Vwijn, dnS, dwijn)

print(cS, dcS)

gewicht = 64.066 * nS #M * n, gewicht in 20mL
dgewicht = 64.066 * dnS
print(gewicht * 50 * 1000, dgewicht * 50 * 1000)
print(Vtit_list)
```
