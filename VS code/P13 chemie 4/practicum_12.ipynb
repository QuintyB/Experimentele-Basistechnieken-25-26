```python
import numpy as np
import sympy as sp
```

```python
#proef 1, berekenendichthied

tot_Volume = 22
tot_dV = 1
massa = 8.24
dm = 0.01
Volume = 10
dV = 0.02

Rho = massa/Volume

dRho = np.sqrt((dm/Volume )**2 + (massa *dV / Volume**2)**2)

display(Rho,r"$/pm$",dRho)
```

```python
#proef 1, Volume procent
VProcentWater = (Rho - 0.8041)/0.0021


dVProcentWater = np.sqrt(((1/0.0021)*dRho)**2)

print (VProcentWater,'+',dVProcentWater)

```

```python
#proef 1, volume ethanol destilaat
VProcentEthanol = 100-VProcentWater
dVProcentEthanol = dVProcentWater

VolumeEthanolDestilaat = tot_Volume * VProcentEthanol/100
dVolumeEthanolDestilaat = np.sqrt((VProcentEthanol*tot_dV/100)**2+(tot_Volume * dVProcentEthanol/100)**2)

print(VolumeEthanolDestilaat,"+",dVolumeEthanolDestilaat)
```

```python
#proef 1, volume procent wijn

volumeWijn = 150
dVolumeWijn = 1
AlcoholProcent = VolumeEthanolDestilaat/volumeWijn
dAlcoholProcent = np.sqrt((1/volumeWijn * dVolumeEthanolDestilaat )**2+(VolumeEthanolDestilaat/(volumeWijn)**2*dVolumeWijn)**2)

print(AlcoholProcent*100,"+",dAlcoholProcent*100)
```
