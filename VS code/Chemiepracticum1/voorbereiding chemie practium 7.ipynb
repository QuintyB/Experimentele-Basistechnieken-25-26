```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
R0 = 1.0        
k = 0.2       
t = np.linspace(0, 25, 500) 
```

```python
R_nulde = R0 - k * t
R_eerste = R0 * np.exp(-k * t)
R_tweede = 1 / (k * t + (1/R0))
```

```python
fig, ax = plt.subplots()
ax.plot(t, R_nulde, color = 'green', label ="m=0")
ax.plot(t, R_eerste, color = "blue" , label ="m=1")
ax.plot(t, R_tweede, color = "orange" , label ="m=2")

ax.set_title("Het verloop van de concentratie in funcie van de tijd")
ax.set_xlabel("Tijd ($t$)")
ax.set_ylabel("concentratie([R])")
ax.legend()
ax.set_ylim(0,1.5)
plt.tight_layout()
plt.show
```
