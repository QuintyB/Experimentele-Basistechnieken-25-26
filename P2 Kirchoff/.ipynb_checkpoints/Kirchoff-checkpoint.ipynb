```python
Vk1 = [4.55, 4.6, 4.61, 4.63, 4.66, 4.68] #V
I1 = [8.07, 5.67, 5.15, 3.89, 2.62, 1.65] #mA

Vk2 = [10.14, 10.42, 10.49, 10.64, 10.80, 10.93]
I2 = [17.99, 12.89, 11.74, 8.95, 6.07, 3.86]

VE1 = 4.71
VE2 = 11.15

fout_Vk1 = []
fout_Vk2 = []
fout_I1 = []
fout_I2 = []

for i in range(len(Vk1)):
    foutV1 = (0.01*Vk1[i] + 0.01)/2
    foutV2 = (0.01*Vk2[i] + 0.01)/2
    foutI1 = (0.01*I1[i] + 0.01)/2
    foutI2 = (0.01*I2[i] + 0.01)/2

    fout_Vk1.append(foutV1)
    fout_Vk2.append(foutV2)
    fout_I1.append(foutI1)
    fout_I2.append(foutI2)

#display(fout_Vk1,fout_Vk2,fout_I1,fout_I2)

fout_Rv1 = []
fout_Rv2 = []
Rv11 = []
Rv22 = []
for i in range(len(Vk1)):
    Rv1 = (VE1 - Vk1[i]) / (I1[i]*10**-3)
    Rv2 = (VE2 - Vk2[i]) / (I2[i]*10**-3)

    Rv11.append(Rv1)
    Rv22.append(Rv2)
    
    sigma1 = sp.sqrt((1/(I1[i]*10**-3))**2 * (0.0571/2)**2 + (1/(I1[i]*10**-3))**2 * fout_Vk1[i]**2 + ((VE1 - Vk1[i]) / (I1[i]*10**-3)**2)**2 * (fout_I1[i]*10**-3)**2)
    sigma2 = sp.sqrt((1/(I2[i]*10**-3))**2 * (0.1215/2)**2 + (1/(I2[i]*10**-3))**2 * fout_Vk2[i]**2 + ((VE2 - Vk2[i]) / (I2[i]*10**-3)**2)**2 * (fout_I2[i]*10**-3)**2)

    fout_Rv1.append(sigma1)
    fout_Rv2.append(sigma2)

#display(Rv11, Rv22)
#display(fout_Rv1, fout_Rv2)

teller = 0
noemer = 0
for i in range(len(Vk1)):
    teller += 1/(fout_Rv1[i]**2) * Rv11[i]
    noemer += 1/(fout_Rv1[i]**2)

gewogenR1 = teller/noemer
display(gewogenR1)

teller = 0
noemer = 0
for i in range(len(Vk1)):
    teller += 1/(fout_Rv2[i]**2) * Rv22[i]
    noemer += 1/(fout_Rv2[i]**2)

gewogenR2 = teller/noemer
display(gewogenR2)

sumg1 = 0
sumg2 = 0
for i in range(len(Vk1)):
    sumg1 += 1/(fout_Rv1[i]**2)
    sumg2 += 1/(fout_Rv2[i]**2)

epsilon1 = 1 / sp.sqrt(sumg1)
epsilon2 = 1 / sp.sqrt(sumg2)

display(2*epsilon1, 2*epsilon2)
```

```python
Vk, I, R1, R2 = sp.symbols("Vk I R1 R2")
Ra = sp.Function("Ra")

Ra = ( 1 / (Vk / I - R1) - 1/R2)**-1

fout_Ra = sp.sqrt((sp.diff(Ra, Vk))**2 * 0.03**2 + (sp.diff(Ra, I))**2 * (0.02*10**-3)**2 + (sp.diff(Ra, R1))**2 * 7.5**2 + (sp.diff(Ra, R2))**2 * 0.01**2)
fout_waarde = fout_Ra.subs([(Vk, 4.6), (I, 2.84*10**-3), (R1, 750), (R2, 10)])
display(fout_waarde)
```

```python
I1 = 6.38e-3
I2 = 3.85e-3
I3 = 2.57e-3

fI1 = 0.01+0.01*I1
fI2 = 0.01+0.01*I2
fI3 = 0.01+0.01*I3

m = 1 / 3 * (I1+I2+I3)

s = sp.sqrt(1/2 * ((I1-m)**2 + (I2-m)**2 + (I3-m)**2))

display(s)

z = (I1-I2-I3) / s
display(z)
```

```python
Vt, In = sp.symbols("Vt In")
Rn = sp.Function("Rn")
Rn = Vt/In

sRn = sp.sqrt((sp.diff(Rn, Vt))**2 * 0.0124**2 + (sp.diff(Rn, In))**2 * (0.0129*10**-3)**2)
sRn_waarde = sRn.subs([(Vt, 1.48), (In, 1.56*10**-3)])
display(sRn_waarde)
```
